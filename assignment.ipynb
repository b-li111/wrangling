```python
import numpy as np
import pandas as pd

airbnb = pd.read_csv("airbnb_hw.csv")
police = pd.read_csv("mn_police_use_of_force.csv")

airbnb['Price_clean'] = (
    airbnb['Price']
    .replace('[\$,]', '', regex=True)
    .astype(str)
    .str.replace(',', '')
)

airbnb['Price_clean'] = pd.to_numeric(airbnb['Price_clean'], errors='coerce')

missing_count = airbnb['Price_clean'].isna().sum()
print("missing values in price_clean:", missing_count)
```

```python
injury_counts = police['subject_injury'].value_counts(dropna=False)
print(injury_counts)
missing_injury = police['subject_injury'].isna().sum()
print("missing values:", missing_injury)

total_rows = police.shape[0]
proportion_missing = missing_injury / total_rows
print("prop missing:", round(proportion_missing, 2))


police['subject_injury_clean'] = police['subject_injury'].fillna('Unknown')

print(police['subject_injury_clean'].value_counts())


injury_vs_force = pd.crosstab(police['subject_injury_clean'], police['force_type'])
print(injury_vs_force)

```
